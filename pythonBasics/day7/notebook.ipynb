```python
#Dimensions of a Numpy array
#to check the dimension of a numpy array we make use of the attribute 'ndim'
# The dimension of an ndarray are also known as 'ranks' of the array

# The kinds of arrays that can be created are :
# Zero Dimensional Array
# One Dimensional Array
# Two Dimensional Array (basically a matrix)
```

```python
# Zero Dimensional Array
import numpy as np
zerodimArray = np.array(46)
print(zerodimArray)
print(type(zerodimArray))
print(f"Dimension :{zerodimArray.ndim}")
```

```python
# one  Dimensional Array
import numpy as np
onedimArray = np.array([1,2,3,4])
print(onedimArray)
print(type(onedimArray))
print(f"Dimension :{onedimArray.ndim}")
```

```python
# Two  Dimensional Array
import numpy as np
# both the array should have the same number of elements reason it is like a matrics so , here in this case it is either 2x4 or 4x2 but for this [[1,2,3,4],[1,2,7]] structure is not defined
twodimArray = np.array([[1,2,3,4],[1,2,7,3]])
print(twodimArray)
print(type(twodimArray))
print(f"Dimension :{twodimArray.ndim}")
```

```python
# Initialzing an array
# To initialize an empty array in numpy we use numpy.zeros() function

import numpy as np

initArray = np.zeros([3,4])
print(initArray)
print(type(initArray))

#Initialize this array with some values using loops
# make use of np.shape()

for i in range (initArray.shape[0]):
  for j in range (initArray.shape[1]):
    initArray[i,j] = (i+1)*(j+1)

print(initArray)
```
