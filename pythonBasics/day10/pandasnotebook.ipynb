```python
'''
pandas dataframes are created by pandas.DataFrame()
this are the form of tabuler structure display of data
parameters included in this dataframe are:
data : the exact data that you are trying to convert into a dataframe
index : gives us the index of a spesified cell(S)
dtype : gives us the datatype of each column
columns : to manipulate column labels
copy : copy of the data into another dataframe
'''
```

```python
import pandas as pd
data = {
    'Student' : ["yug","ananya","divya","kanishka","anvita","annu","shivika","om sharma"],
    'subject' : ["math","science","english","hindi","social","computer","biology","chemistry"],
    'Marks' : [90,80,70,60,50,40,30,20]
}
dataframe = pd.DataFrame(data)
print(dataframe)

dataframe1 = pd.DataFrame(data,index=["index1","index2","index3","index4","index5","index6","index7","index8"])
print(dataframe1)

# accessing these elements
print(dataframe1.loc["index1","Student"]) #
print(dataframe1.iloc[[2,3]]) #using index

#accessing columns
for col in dataframe1:
    print(col)
```

```python
# pandas dataframe attributes and methods
'''
dtype
index
size
shape
T - transpose(2,3)->(3,2)
head()
tail()
'''
data = {
    'Student' : ["yug","ananya","divya","kanishka","anvita","annu","shivika","om sharma"],
    'subject' : ["math","science","english","hindi","social","computer","biology","chemistry"],
    'Marks' : [90,80,70,60,50,40,30,20],
    'Grade' : ["A","B","C","D","E","F","G","H"],
    'Age' : [20,20,20,20,20,20,20,20],
    'Gender' : ["male","female","female","female","female","male","female","male"]
}
dataframe = pd.DataFrame(data)
print(dataframe)
print("dtype---------------------")
print(dataframe.dtypes)
print("index---------------------")
print(dataframe.index)
print("size---------------------")
print(dataframe.size)
print("shape---------------------")
print(dataframe.shape)
print("transpose---------------------")
print(dataframe.T)
print("head---------------------")
print(dataframe.head(2))
print("tail---------------------")
print(dataframe.tail(3))
```
