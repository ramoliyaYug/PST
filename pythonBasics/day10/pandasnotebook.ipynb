```python
'''
pandas dataframes are created by pandas.DataFrame()
this are the form of tabuler structure display of data
parameters included in this dataframe are:
data : the exact data that you are trying to convert into a dataframe
index : gives us the index of a spesified cell(S)
dtype : gives us the datatype of each column
columns : to manipulate column labels
copy : copy of the data into another dataframe
'''
```

```python
import pandas as pd
data = {
    'Student' : ["yug","ananya","divya","kanishka","anvita","annu","shivika","om sharma"],
    'subject' : ["math","science","english","hindi","social","computer","biology","chemistry"],
    'Marks' : [90,80,70,60,50,40,30,20]
}
dataframe = pd.DataFrame(data)
print(dataframe)

dataframe1 = pd.DataFrame(data,index=["index1","index2","index3","index4","index5","index6","index7","index8"])
print(dataframe1)

# accessing these elements
print(dataframe1.loc["index1","Student"]) #
print(dataframe1.iloc[[2,3]]) #using index

#accessing columns
for col in dataframe1:
    print(col)
```

```python
# pandas dataframe attributes and methods
'''
dtype
index
size
shape
T - transpose(2,3)->(3,2)
head()
tail()
'''
data = {
    'Student' : ["yug","ananya","divya","kanishka","anvita","annu","shivika","om sharma"],
    'subject' : ["math","science","english","hindi","social","computer","biology","chemistry"],
    'Marks' : [90,80,70,60,50,40,30,20],
    'Grade' : ["A","B","C","D","E","F","G","H"],
    'Age' : [20,20,20,20,20,20,20,20],
    'Gender' : ["male","female","female","female","female","male","female","male"]
}
dataframe = pd.DataFrame(data)
print(dataframe)
print("dtype---------------------")
print(dataframe.dtypes)
print("index---------------------")
print(dataframe.index)
print("size---------------------")
print(dataframe.size)
print("shape---------------------")
print(dataframe.shape)
print("transpose---------------------")
print(dataframe.T)
print("head---------------------")
print(dataframe.head(2))
print("tail---------------------")
print(dataframe.tail(3))
```

```python
import pandas as pd
data1 = {
    'Student' : ["yug","ananya","divya","kanishka","anvita"],
    'subject' : ["math","science","english","hindi","social"],
    'Marks' : [90,80,70,60,50]
}
data2 = {
    "student" : ["annu","shivika","om sharma","sneha","riya"],
    "subject" : ["computer","biology","chemistry","math","social"],
    "marks" : [80,70,60,50,40]
}
dataframe1 = pd.DataFrame(data1)
dataframe2 = pd.DataFrame(data2)
# joining of dataframes
print(dataframe1.join(dataframe2, lsuffix="_left", rsuffix="_right"))
#concate method
print(pd.concat([dataframe1,dataframe2]))
```

```python
    # pandas series
'''
pandas series is kind of one dimenstinal array which stores the data in a column like structure with indexes
we make use of the series() method to create the pandas seties

patameters /arguments of the series method
data
index
column
copy
dtype
name
index_col
'''
import pandas as pd
data = [10,20,30,40,50,60]
series = pd.Series(data)
print(series)
print(series[3])
```

```python
# pandas attributes and methods
'''
dtype
ndim
size
shape
index
hasnans

head()
tail()
info()
'''

data1 = pd.Series([10,20,30,40,50,60,"hello","hi"])
print(data1)
print("dtype---------------------")
print(data1.dtype)
print("ndim---------------------")
print(data1.ndim)
print("shape---------------------")
print(data1.shape)
print("size---------------------")
print(data1.size)
print("index---------------------")
print(data1.index)
print("hasnans---------------------")
print(data1.hasnans)
print("head---------------------")
print(data1.head(2))
print("tail----------------------")
print(data1.tail(3))
print(data1.tail(2))
print("info----------------------")
print(data1.info())
```
