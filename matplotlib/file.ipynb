```python
# Displaying data through a pandas Dataframe

import matplotlib.pyplot as plt
import pandas as pd

data = {
    "Month" : ["Jan", "Feb", "Mar", "April", "Jun"],
    "Sales" : [150,200,180,250,300]
}
df = pd.DataFrame(data)
plt.plot(df["Month"], df["Sales"],marker = "o", linestyle = "--", color= "purple")

plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales(per unit)")

plt.grid(True)

plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

arr1 = np.arange(5)
arr2 = np.array([2,4,6,8,10])
arr3 = np.array([5,6,7,8,9])

fig = plt.figure()
ax = plt.subplot()

# Plotting two different subplots(line graphs)
ax.plot(arr1,arr2, "g--", label = "First_launch")
ax.plot(arr1,arr3, "r:", label = "Second_launch")

#Adding the legend

plt.legend(loc = "lower right")
plt.show()

#(upper left, upper right, upper center, lower left, lower right, lower center, center)
```

```python
import matplotlib.pyplot as plt
import numpy as np

days = np.array([1, 2, 3, 4, 5])
temp = np.array([30, 32, 35, 33, 36])
plt.plot(days, temp, marker = "o", linestyle = "--", color= "orange", label = "Temp in C")
plt.xlabel("Days")
plt.ylabel("Temperature")
plt.title("Temperature vs Days")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

days = np.array(['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'])
rainfall = np.array([12, 8, 15, 7, 10, 6, 9])

"""Use square markers 's'
Line color: blue
Line style: dashed '--'
Title: “Weekly Rainfall”
X-label: “Day of Week”
Y-label: “Rainfall (mm)”
"""

plt.plot(days, rainfall, marker = "s", linestyle = "--", color= "blue", label = "Rainfall in mm")
plt.xlabel("Day of Week")
plt.ylabel("Rainfall (mm)")
plt.title("Weekly Rainfall")
plt.legend()
plt.show()
```
